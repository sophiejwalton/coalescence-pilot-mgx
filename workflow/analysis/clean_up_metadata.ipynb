```python
import pandas as pd
```

```python
df = pd.read_csv('e003_coalescence_metadata_round4_good.csv')
columns = df.columns.values
bad_columns = []
for col in columns:
    if 'Unnamed:' in col:
        bad_columns.append(col)
df = df.drop(columns = bad_columns)
df['contaminated_subject'] = df['parent_subjects'].transform(lambda x: 'AC' in x)
df.to_csv('TableS3_community_coalescence_metadata.csv',index=False)
```
