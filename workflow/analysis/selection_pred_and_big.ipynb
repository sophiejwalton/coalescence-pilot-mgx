```python
def get_neutral_bootstrap_drift(f0):
    # returns delta f 
    Ne = (1e9/5)*1/10 #(typical Ne for sp at 10 %) 
    dt = np.log2(200)*7 # 7 passages 1/200 dln 
    Zt = np.random.normal()
    drift = (f0*(1-f0)*dt/Ne)**(1/2)
    return drift# (drift is delta f), 

def get_sel_from_bootstrap(f0,drift):
    f0_coeff = (1-f0)/(f0)
    f7=f0+drift
    dt = np.log2(200)*7
    p7_s = 1/dt*(np.log(f7/(1-f7))+np.log(f0_coeff))
    return p7_s

def get_bootstrap_reps(f0s,reps = 10):
    all_reps = np.zeros((reps,len(f0s)))
    for i,rep in enumerate(np.arange(reps)):
        f0_bs = np.random.choice(f0s, size=len(f0s))
        drifts = get_neutral_bootstrap_drift(f0_bs)
        sels = get_sel_from_bootstrap(f0_bs,drifts)
        all_reps[i, :] = sels
    return all_reps

        
        
```
