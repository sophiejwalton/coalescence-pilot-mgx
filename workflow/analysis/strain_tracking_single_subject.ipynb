```python
cmap = {'fecal': bokeh.palettes.Set2[8][-2],
        'mBHI':bokeh.palettes.Set2[8][1],
        'mGAM': bokeh.palettes.Set2[8][0],}
```

```python
def plot_trajectories_with(freq_df, minor_strain, separate_plots_per_mesocosm=True):
    p2 = bokeh.plotting.figure(width = 500, height = 300)
   # palette = bokeh.palettes.colorblind['Colorblind'][7][::-1]
    for i, type_meso in enumerate(freq_df['type_mesocosm'].unique()):
        mesos = freq_df.loc[freq_df['type_mesocosm'] ==type_meso, 'mesocosm'].unique()
        media = type_meso.split('-')[1]
        if 'fecal' in type_meso:
            media = 'fecal'
   
      #  env = f'{media1}-{media2}'
        color = cmap[media]
        for mesocosm in mesos:
            print(mesocosm)
            df_meso = freq_df.loc[freq_df['mesocosm'] == mesocosm,:]
            in_sample = df_meso['inoculumn_sample'].unique()[0]
           # if in_sample not in df_meso.index.values:
            #    df_meso.loc[in_sample,:] = pd.concat([df_meso,freq_df.loc[in_sample,:]],axis=1)
                
            df_meso = df_meso.sort_values(by = 'passage')
                
            p2.line(df_meso['passage'].values,
                   df_meso[minor_strain].values,color=color)
            p2.circle(df_meso['passage'].values,
                   df_meso[minor_strain].values,color = color, legend_label = type_meso, size = 5)
                           
    p2.yaxis.axis_label = f'Frequency'
    p2.xaxis.axis_label = 'Passage'
    inoculumn_stuff = '-'.join(type_meso.split('-')[:-1])
  #  p2.title.text = f'{minor_strain_subject} in {inoculumn_stuff}'
    p2.y_range = bokeh.models.Range1d(-.005,1.005)
   # p2.x_range = bokeh.models.Range1d(-.5,7.5)
    p2.xgrid.grid_line_color = None
    p2.ygrid.grid_line_color = None

    return p2

def plot_trajectories_with_in(freq_df, minor_strain, separate_plots_per_mesocosm=True):
    p2 = bokeh.plotting.figure(width = 500, height = 300)
   # palette = bokeh.palettes.colorblind['Colorblind'][7][::-1]
    for i, type_meso in enumerate(freq_df['type_mesocosm'].unique()):
        mesos = freq_df.loc[freq_df['type_mesocosm'] ==type_meso, 'mesocosm'].unique()
      #  print(mesos)
        media = type_meso.split('-')[1]
        if 'fecal' in type_meso:
          #  print(type_meso, 'fecal')
            media = 'fecal'
   
      #  env = f'{media1}-{media2}'
        color = cmap[media]
        for mesocosm in mesos:
           
            df_meso = freq_df.loc[freq_df['mesocosm'] == mesocosm,:]
          #  print('shee',df_meso)
            in_sample = df_meso['inoculumn_sample'].unique()[0]
            if 'fecal' not in mesocosm:
                if in_sample not in df_meso['sample'].values:
                 #   print('yay')
                    df_meso= pd.concat([df_meso,freq_df.loc[freq_df['sample'] == in_sample,:]],axis=0)
                
           # print('woo', df_meso.columns.values)
            df_meso = df_meso.sort_values(by = 'passage')
                
            p2.line(df_meso['passage'].values,
                   df_meso[minor_strain].values,color=color)
            p2.circle(df_meso['passage'].values,
                   df_meso[minor_strain].values,color = color, legend_label = type_meso, size = 5)
                           
    p2.yaxis.axis_label = f'Frequency'
    p2.xaxis.axis_label = 'Passage'
   # inoculumn_stuff = '-'.join(type_meso.split('-')[:-1])
  #  p2.title.text = f'{minor_strain_subject} in {inoculumn_stuff}'
    p2.y_range = bokeh.models.Range1d(-.005,1.005)
   # p2.x_range = bokeh.models.Range1d(-.5,7.5)
    p2.xgrid.grid_line_color = None
    p2.ygrid.grid_line_color = None

    return p2
```

```python


def make_plots(species, subject):
    fname = f'/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/track_snpsv2_ALL_same_subject/{species}/{subject}_parent_freqs.csv'
    df = pd.read_csv(fname).rename(columns = {'Unnamed: 0': 'sample'}).set_index('sample')

    in_sample1 = df.columns.values[0]
    in_sample2 = df.columns.values[1]
    df_sum = df.sum(axis=1)
    good1=df_sum[df_sum>.15]
    good1 = good1[good1<1.15]
    df = df.loc[good1.index.values,:].reset_index()
    #df = df.reset_index()
    
    if len(df)==0:
        return []
    assembly_metadata_AF = assembly_all.loc[assembly_all['subject']==subject,:].reset_index()
    df_AE = df.loc[df['sample'].isin(assembly_metadata_AF['sample'].unique()),:]
    df_AE['mesocosm'] = df_AE['sample'].transform(lambda x: assembly_metadata_AF.loc[assembly_metadata_AF['sample']==x,'mesocosm'].values[0])
    df_AE['type_mesocosm'] = df_AE['sample'].transform(lambda x: assembly_metadata_AF.loc[assembly_metadata_AF['sample']==x,'type_mesocosm'].values[0])
    df_AE['pseudo_time_passage'] = df_AE['sample'].transform(lambda x: 
                                                             int(assembly_metadata_AF.loc[assembly_metadata_AF['sample']==x,'pseudo_time_passage'].values[0]))
    df_AE['inoculumn_sample'] = df_AE['sample'].transform(lambda x: 
                                                             assembly_metadata_AF.loc[assembly_metadata_AF['sample']==x,
                                                                 'inoculumn_sample'].values[0])

    df_AE['passage'] = df_AE['pseudo_time_passage']
   # print(df_AE)
#    print(
    if len(df_AE)==0:
        return []
   # print(df_AE['media'].unique())
    fecal = list(assembly_metadata_AF.loc[assembly_metadata_AF['mesocosm'] == f'fecal-{subject}-fecal','sample'].values)
    
   # print(sp, fecal[0] in df_AE['sample'].values)
    good_in = list(df_AE.loc[df_AE['inoculumn_sample'] == in_sample1,'sample'].values) + fecal
   # print(good_in)
    p1 = plot_trajectories_with_in(df_AE.loc[df_AE['sample'].isin(good_in),:],
                               in_sample1, separate_plots_per_mesocosm=True)
    p1.legend.visible=False
    p1.title.text = f'{subject} in {species} in mBHI'
   # bokeh.io.show(p1)
    good_in = list(df_AE.loc[df_AE['inoculumn_sample'] == in_sample2,'sample'].values) + fecal
    p2= plot_trajectories_with_in(df_AE.loc[df_AE['sample'].isin(good_in),:],
                               in_sample2, separate_plots_per_mesocosm=True)
    p2.title.text = f'{subject} in {species} in mGAM'
    p2.legend.visible=False
  #  bokeh.io.show(p2)
    return [p1,p2]
```
