```python
df_abundance = pd.read_csv('e003_round4_assembly_abundances.csv')
df_abundance_coalescence = pd.read_csv('e004_abundances_no_dups.csv')
df_abundance_coalescence['exp_type'].unique()
df_abundance.loc[df_abundance['exp_type']=='e003GlycerolRevival','passage']=df_abundance.loc[df_abundance['exp_type']=='e003GlycerolRevival','passage']+5
```

```python
df_abundance['counts']=1.
df_abundance['no_zero']=(df_abundance['relative_abundance']>0.)
df_abundancegr=df_abundance.groupby(['sample']).sum(numeric_only=True)
df_abundancegr = df_abundancegr.loc[df_abundancegr['relative_abundance']>0,:]
df_abundancegr =df_abundancegr.loc[df_abundancegr['no_zero']>10,:]
df_abundancegr['no_zero'].min()
good_samples = df_abundancegr.index.values
df_abundance=df_abundance.loc[df_abundance['sample'].isin(good_samples),:]
```

```python
meso_to_look_at = 'B8-AA-mGAM'
for meso_to_look_at in df_abundance['mesocosm'].unique():
    if 'AC' in meso_to_look_at:
        continue
    bar_width = 1.0
    p = make_bar_plot_assembly(meso_to_look_at, df_abundance,bar_width = bar_width, 
                              include_glycerol=True)
    p.output_backend = "svg"
    fname_prefix = f'{meso_to_look_at}_bar_plot_{str(bar_width*10)}'
    tick_font_size ='15px'
    label_font_size = '22px'
    p.yaxis.axis_label_text_font_style = 'normal'
    p.xaxis.axis_label_text_font_style = 'normal'
    p.xaxis.axis_label_text_font_size=label_font_size
    p.xaxis.major_label_text_font_size=tick_font_size
    p.yaxis.axis_label_text_font_size=label_font_size
    p.yaxis.major_label_text_font_size=tick_font_size
    p.xaxis.axis_label = 'Timepoint'
    p.xaxis.minor_tick_line_color= None
    p.yaxis.minor_tick_line_color= None
    p.title.text = meso_to_look_at
    p.title.text_font_style='normal'
    p.title.text_font_size=label_font_size

    export_plot_pdf(p, fname_prefix)
    bokeh.io.show(p)
```

```python
def ecdf_transform_no_norm(data):
    return data.rank(method="first",ascending=False) #/ len(data)

def make_rank_abundance(type_mesocosm):
    if type_mesocosm.split('-')[-1]=='mGAM':
        color = bokeh.palettes.Set2[8][0]
    if type_mesocosm.split('-')[-1]=='mBHI':
        color = bokeh.palettes.Set2[8][1]
    
    AAmGAM = df_abundance.loc[(df_abundance['type_mesocosm']==type_mesocosm)*(df_abundance['passage']==5),:]
    AAmGAM = AAmGAM.loc[AAmGAM['relative_abundance']>0,:]
    AAmGAM["rel abun ECDF"] = AAmGAM.groupby("sample")[
        "relative_abundance"
    ].transform(ecdf_transform_no_norm)
    
    mesos = AAmGAM['mesocosm'].unique()
    ps = []
    for i,meso in enumerate(mesos):
        if 'fecal' in mesos:
            continue
        if i == 0:
            p1 = hv.Curve(
                data=AAmGAM.loc[AAmGAM['mesocosm']==meso,:].sort_values(by='rel abun ECDF'),
                vdims=['relative_abundance','sample'],
                kdims=[('rel abun ECDF', 'ECDF'), ],label=type_mesocosm.split('-')[-1],
            ).opts(width=200,height=200, color=  color,#xlabel='Relative Abundance',
                   logy=True,logx=True,)
        else:
            p1 = hv.Curve(
                data=AAmGAM.loc[AAmGAM['mesocosm']==meso,:].sort_values(by='rel abun ECDF'),
                vdims=['relative_abundance','sample'],
                kdims=[('rel abun ECDF', 'ECDF'), ],
            ).opts(width=200,height=200, color=  color,#xlabel='Relative Abundance',
                   logy=True,logx=True,)
        ps.append(p1)
        #  ylabel='1-P(f>x)')


    sub = type_mesocosm.split('-')[0]
    AAmGAM = df_abundance.loc[df_abundance['type_mesocosm']==f'{sub}-fecal',:]
    AAmGAM = AAmGAM.loc[AAmGAM['relative_abundance']>0,:]
    AAmGAM["rel abun ECDF"] = AAmGAM.groupby("sample")[
        "relative_abundance"
    ].transform(ecdf_transform_no_norm)

    p4 = hv.Curve(
        data=AAmGAM.sort_values(by='rel abun ECDF'),
        vdims=['relative_abundance','sample'],
        kdims=[('rel abun ECDF', 'ECDF'), ],label='Fecal',
    ).opts(width=250,height=400, color= bokeh.palettes.Set2[8][-2],#xlabel='Relative Abundance',
           logy=True,logx=True,ylim=(1e-3,1))
        #  ylabel='1-P(f>x)')
    if len(ps)==3:
        p=hv.render(p4*ps[0]*ps[1]*ps[2])
    elif len(ps)==2:
        p=hv.render(p4*ps[0]*ps[1])
    elif len(ps)==1:
        p=hv.render(p4*ps[0])
    else:
        p = hv.render(p4)
    tick_font_size ='15px'
    label_font_size = '22px'
    p.yaxis.axis_label_text_font_style = 'normal'
    p.xaxis.axis_label_text_font_style = 'normal'
    p.xaxis.axis_label_text_font_size=label_font_size
    p.xaxis.major_label_text_font_size=tick_font_size
    p.yaxis.axis_label_text_font_size=label_font_size
    p.yaxis.major_label_text_font_size=tick_font_size
    p.xaxis.axis_label = 'Species Rank'
    p.yaxis.axis_label = 'Relative Abundance'
    p.xaxis.minor_tick_line_color= None
    p.yaxis.minor_tick_line_color= None
    p.title.text = type_mesocosm
    p.title.text_font_style='normal'
    p.title.text_font_size=label_font_size
 #   bokeh.io.show(p)
    p.output_backend='svg'
    export_plot_pdf(p,f'rank_abundance_{type_mesocosm}')
    return p 
```

```python
ps_all = []
for meso in ['AA-mBHI','AE-mBHI','AF-mBHI','AA-mGAM','AE-mGAM','AF-mGAM',]:
    ps = make_rank_abundance(meso)
    ps_all.append(ps)
```

```python
grid = bokeh.layouts.gridplot(ps_all,ncols=3)
bokeh.io.show(grid)
fname_prefix = 'all_rank_rels'
plot_fname_svg=f'plots/{fname_prefix}.svg'
plot_fname_pdf=f'plots/{fname_prefix}.pdf'
bokeh.io.export_svg(grid,filename=plot_fname_svg)
convert_scripts = f'rsvg-convert -f pdf -o {plot_fname_pdf} {plot_fname_svg}' 
os.system(convert_scripts)
```
