```python
def ecdf_transform_no_norm(data):
    return data.rank(method="first",ascending=False) #/ len(data)

def make_rank_abundance(type_mesocosm):
    if type_mesocosm.split('-')[-1]=='mGAM':
        color = bokeh.palettes.Set2[8][0]
    if type_mesocosm.split('-')[-1]=='mBHI':
        color = bokeh.palettes.Set2[8][1]
    
    AAmGAM = df_abundance.loc[(df_abundance['type_mesocosm']==type_mesocosm)*(df_abundance['passage']==5),:]
    AAmGAM = AAmGAM.loc[AAmGAM['relative_abundance']>0,:]
    AAmGAM["rel abun ECDF"] = AAmGAM.groupby("sample")[
        "relative_abundance"
    ].transform(ecdf_transform_no_norm)
    
    mesos = AAmGAM['mesocosm'].unique()
    ps = []
    for i,meso in enumerate(mesos):
        if 'fecal' in mesos:
            continue
        if i == 0:
            p1 = hv.Curve(
                data=AAmGAM.loc[AAmGAM['mesocosm']==meso,:].sort_values(by='rel abun ECDF'),
                vdims=['relative_abundance','sample'],
                kdims=[('rel abun ECDF', 'ECDF'), ],label=type_mesocosm.split('-')[-1],
            ).opts(width=200,height=200, color=  color,#xlabel='Relative Abundance',
                   logy=True,logx=True,)
        else:
            p1 = hv.Curve(
                data=AAmGAM.loc[AAmGAM['mesocosm']==meso,:].sort_values(by='rel abun ECDF'),
                vdims=['relative_abundance','sample'],
                kdims=[('rel abun ECDF', 'ECDF'), ],
            ).opts(width=200,height=200, color=  color,#xlabel='Relative Abundance',
                   logy=True,logx=True,)
        ps.append(p1)
        #  ylabel='1-P(f>x)')


    sub = type_mesocosm.split('-')[0]
    AAmGAM = df_abundance.loc[df_abundance['type_mesocosm']==f'{sub}-fecal',:]
    AAmGAM = AAmGAM.loc[AAmGAM['relative_abundance']>0,:]
    AAmGAM["rel abun ECDF"] = AAmGAM.groupby("sample")[
        "relative_abundance"
    ].transform(ecdf_transform_no_norm)

    p4 = hv.Curve(
        data=AAmGAM.sort_values(by='rel abun ECDF'),
        vdims=['relative_abundance','sample'],
        kdims=[('rel abun ECDF', 'ECDF'), ],label='Fecal',
    ).opts(width=250,height=400, color= bokeh.palettes.Set2[8][-2],#xlabel='Relative Abundance',
           logy=True,logx=True,ylim=(1e-3,1))
        #  ylabel='1-P(f>x)')
    if len(ps)==3:
        p=hv.render(p4*ps[0]*ps[1]*ps[2])
    elif len(ps)==2:
        p=hv.render(p4*ps[0]*ps[1])
    elif len(ps)==1:
        p=hv.render(p4*ps[0])
    else:
        p = hv.render(p4)
    tick_font_size ='15px'
    label_font_size = '22px'
    p.yaxis.axis_label_text_font_style = 'normal'
    p.xaxis.axis_label_text_font_style = 'normal'
    p.xaxis.axis_label_text_font_size=label_font_size
    p.xaxis.major_label_text_font_size=tick_font_size
    p.yaxis.axis_label_text_font_size=label_font_size
    p.yaxis.major_label_text_font_size=tick_font_size
    p.xaxis.axis_label = 'Species Rank'
    p.yaxis.axis_label = 'Relative Abundance'
    p.xaxis.minor_tick_line_color= None
    p.yaxis.minor_tick_line_color= None
    p.title.text = type_mesocosm
    p.title.text_font_style='normal'
    p.title.text_font_size=label_font_size
 #   bokeh.io.show(p)
    p.output_backend='svg'
    export_plot_pdf(p,f'rank_abundance_{type_mesocosm}')
    return p 
```

```python
ps_all = []
for meso in ['AA-mBHI','AE-mBHI','AF-mBHI','AA-mGAM','AE-mGAM','AF-mGAM',]:
    ps = make_rank_abundance(meso)
    ps_all.append(ps)
```
