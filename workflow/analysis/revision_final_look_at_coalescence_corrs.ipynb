```python
def calculate_qvalues(pvalues):

    # Uses the formula
    #
    # Qi = min_{Q>Pi} { Q*(sum_j 1)/(sum_j theta(Q-Pj)) } 
    #
    # Qi = max_{P_(k)>Pi} 

    # calculate q-values
    qvalues = []
    
    sorted_pvalues = np.array(sorted(pvalues))

    Ntot = len(sorted_pvalues)
    Nless =np.array([(sorted_pvalues<=p).sum() for p in sorted_pvalues])

    for p in pvalues:

        min_q = 1e06
            
        for j in reversed(range(0,Ntot)):
             
            if sorted_pvalues[j]<p:
                break
            else:
                new_q = Ntot*sorted_pvalues[j]*1.0/Nless[j]
                if new_q<min_q:
                    min_q = new_q
                    
        qvalues.append(min_q)
    
    qvalues = np.array(qvalues)
    
    return qvalues
```

```python

def boot_gstats(gstats, bootstraps = 10000):
    meds = np.zeros(bootstraps)
    for b in range(bootstraps):
       # print(b)
        new_gstats= np.random.choice(gstats, size = len(gstats))
        meds[b] = np.mean(new_gstats)
    return meds
```

```python
def bootstrap_bias_stats_binomial(ns, bsreps = 100):
    gstat_meds = np.zeros(bsreps)
    for bsrep in range(bsreps):
        rv = np.random.binomial(ns,1/2)
        gstat =  get_Gstat(rv, ns)
        gstat[rv==0] = 2*np.log(2)*ns[rv==0]
        gstat[rv==ns] = 2*np.log(2)*ns[rv==ns]
        gstat_meds[bsrep]=np.median(gstat)
    return gstat_meds

def bootstrap_bias_stats_binomial_pmeds(ns, bsreps = 100):
    p_meds = np.zeros(bsreps)
    for bsrep in range(bsreps):
        rv = np.random.binomial(ns,1/2)
        p = rv/ns
        p_meds[bsrep]=np.median(p)
    return p_meds
```

```python
def get_bsreps_ind(counts,bsreps = 10000):
    reps = np.zeros(bsreps)
    cs = np.zeros(bsreps)
    for i in range(bsreps):
        j = np.random.binomial(counts,1/2)
        cs[i]=j
        g = get_Gstat(j,counts)
        if int(j) == 0.:
            g = 2*counts*np.log(2)
        elif int(j) == counts:
       #     print('o')
            g = 2*counts*np.log(2)
        reps[i]=g

    return reps,cs
```

```python
def bootstrap_bias_stats_binomial_all(ns, bsreps = 1000):
    gstats_all = np.zeros((bsreps, len(ns)))
    for bsrep in range(bsreps):
        rv = np.random.binomial(ns,1/2)
        gstat =  get_Gstat(rv, ns)
        gstat[rv==0] = 2*np.log(2)*ns[rv==0]
        gstat[rv==ns] = 2*np.log(2)*ns[rv==ns]
        gstats_all[bsrep,:]=gstat
    return gstats_all
```

```python
def boot_strap_JSD(fractions, JSDs, bootstraps = 1000):
    rs = np.zeros(bootstraps)
    for b in range(bootstraps):
       # print(b)
        new_JSDs = np.copy(JSDs)
        new_dirs = np.random.choice([1,-1], size=len(new_JSDs))
        new_JSDs = new_JSDs*new_dirs
        new_corr_coeff = pearsonr(fractions, new_JSDs)[0]
        rs[b] = new_corr_coeff
    return rs
        
```
