```python
read_counts = read_Counts[df_abun_get_alpha['sample'].unique()]
read_countsdf = pd.DataFrame(read_counts).reset_index().rename(columns={'index':'sample',0:'read_counts'})
read_countsdf['alpha']=read_countsdf['sample'].transform(lambda x: df_abun_get_alpha.loc[df_abun_get_alpha['sample']==x,'counts'].values[0])
```
