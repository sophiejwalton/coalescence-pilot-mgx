```python
folders = glob('/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/calculateDiversityDepthv3/*')
all_dfs = []
for folder in folders:
    species_id = folder.split('/')[-1]
    fname1 = f'{folder}/{species_id}_med_depth_df.csv'
    df_depth = pd.read_csv(fname1).set_index('Unnamed: 0').rename(columns={'0': 'depth'})

    fname2 = f'{folder}/{species_id}_num_int_sites2.csv'
    df_num_int= pd.read_csv(fname2).set_index('Unnamed: 0').rename(columns={'0': 'num_int_sites'})

    fname3 = f'{folder}/{species_id}_diversity_df2.csv'
    df_div= pd.read_csv(fname3).set_index('Unnamed: 0').rename(columns={'0': 'diversity'})

    full_df = pd.concat([df_depth, df_num_int, df_div],axis=1)
    full_df['species_id'] = int(species_id)
    #full_df['species'] = df_metadata.loc[int(species_id), 'species'].split('s__')[-1]
    if int(species_id) in species_to_use:
        all_dfs.append(full_df)
    
    all_dfs.append(full_df)

    
all_dfs = pd.concat(all_dfs).reset_index().rename(columns = {'Unnamed: 0':'sample'})
#all_dfs_assembly.columns.values
all_dfs['sites'] = all_dfs['num_int_sites']*1/all_dfs['diversity']
all_dfs = all_dfs.loc[~all_dfs['sites'].isna(),:]
all_dfsgr = all_dfs.groupby(['species_id']).max(numeric_only=True)
print(all_dfs['species_id'].unique())
def ecdf_transform(data):
    return 1- data.rank(method="first") / len(data)

all_dfsgr .loc[:, "sites ECDF"] = all_dfsgr[
    "sites"
].transform(ecdf_transform)

p1=hv.Scatter(
    data=all_dfsgr ,
    kdims='sites',
    vdims=[('sites ECDF', 'ECDF'),],
).opts(logx=True,xlim=(1e5,1e7))
p1
```

```python
all_dfs['frac_sites']=np.nan
for spid in all_dfsgr.index.values:
    total_sites = all_dfsgr.loc[spid,'sites']
    all_dfs.loc[all_dfs['species_id'] == spid,'frac_sites'] = all_dfs.loc[all_dfs['species_id'] == spid,'sites']/total_sites 

all_dfs .loc[:, "frac_sites ECDF"] = all_dfs[
    "frac_sites"
].transform(ecdf_transform)

p1=hv.Scatter(
    data=all_dfs ,
    kdims='frac_sites',
    vdims=[('frac_sites ECDF', 'ECDF'),],
).opts()
p1
```
