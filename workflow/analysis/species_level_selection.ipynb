```python
#df_abun_p07['sel_coeff'] = (1/dt)*(np.log(df_abun_p07['p0 relative_abundance']/(1-df_abun_p07['p0 relative_abundance'])) + \
 #                            np.log((1-df_abun_p07['relative_abundance'])/(df_abun_p07['relative_abundance'])))

df_abun_p07['shift_07'] = (np.log(df_abun_p07['p0 relative_abundance']/(1-df_abun_p07['p0 relative_abundance'])) + \
                             np.log((1-df_abun_p07['relative_abundance'])/(df_abun_p07['relative_abundance'])))
df_abun_p07['shift_07']
```

```python
sp_type_mesos = []
diffs = []
m1s = []
m2s = []
for sp_type_meso in df_abun_p07['sp-type_meso']:
    df_small = df_abun_p07.loc[df_abun_p07['sp-type_meso'] == sp_type_meso,:]
    for m1,m2 in it.combinations(df_small['sp-meso'].unique(),2):
        m1_abun = df_small.loc[df_small['sp-meso']==m1,'relative_abundance'].values[0]
        m2_abun = df_small.loc[df_small['sp-meso']==m2,'relative_abundance'].values[0]
        sp_type_mesos.append(sp_type_meso)
        diffs.append((np.log(m1_abun/(1-m1_abun)) + np.log((1-m2_abun)/(m2_abun))))
        m1s.append(m1_abun)
        m2s.append(m2_abun)
    
    
```

```python
df_full_compares = pd.DataFrame(data = {'sp-type_meso':sp_type_mesos,
                                        'diffs':diffs, 'm1s': m1s, 'm2s':m2s,})
df_full_compares  = df_full_compares.loc[(df_full_compares['m1s']>9e-4)*(df_full_compares['m2s']>9e-4),:]
df_full_compares['m2s']
```

```python
def ecdf_transform(data):
    return 1- data.rank(method="first") / len(data)

df_full_compares['diffs_abs']=df_full_compares['diffs'].abs()
df_full_compares.loc[:, "btw rep ECDF"] = df_full_compares[
    "diffs_abs"
].transform(ecdf_transform)
df_abun_p07good = df_abun_p07.loc[df_abun_p07['p0 relative_abundance']>9e-4,:]
df_abun_p07good['shift_07_abs'] = df_abun_p07good['shift_07'].abs()
df_abun_p07good.loc[:, "ovr time ECDF"] = df_abun_p07good[
    "shift_07_abs"
].transform(ecdf_transform)
df_abun_p07
```
