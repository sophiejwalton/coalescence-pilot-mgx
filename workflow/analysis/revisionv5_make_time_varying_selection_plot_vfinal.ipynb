```python
def polarize(df,positive=False):
    to_polarize = df.loc[df['pred7_med']>.5,:].index.values
    if positive:
        to_polarize = df.loc[df['pred7_med']<.5,:].index.values
    print(to_polarize)
    for freq_column in ['act_7_med', 'min_act_7_med',  'max_act_7_med', 'min_pred7_med','max_pred7_med',]:
        df.loc[to_polarize,freq_column]= 1- df.loc[to_polarize,freq_column]
    for sel_column in ['early_sel_med','max_early_sel_med','min_early_sel_med',
                      'late_sel_med','max_late_sel_med','min_late_sel_med',]:
        df.loc[to_polarize,sel_column]= -df.loc[to_polarize,sel_column]
    print(df.loc[to_polarize,'pred7_med'])
    df.loc[to_polarize,'pred7_med']=1-df.loc[to_polarize,'pred7_med']
    return df


def adjust(df,thresh=5e-3):
   # to_polarize = df.loc[df['pred_p7s']>.5,:].index.values
    for freq_column in ['act_7_med', 'min_act_7_med',  'max_act_7_med', 'min_pred7_med','max_pred7_med','pred7_med']:
        df.loc[df[freq_column]>1-thresh,freq_column]= 1-thresh
        df.loc[df[freq_column]<thresh,freq_column]= thresh
    return df
 
    
```
