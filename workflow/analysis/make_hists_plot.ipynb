```python
import pandas as pd
import numpy as np

from snp_analysis_tools_sherlock import *

import iqplot
import bokeh.plotting
import bokeh.io
import holoviews as hv
from holoviews import dim, opts
import bokeh.models
from bokeh.layouts import gridplot
from coalescence_analysis_tools import * 
hv.extension('matplotlib')
hv.extension('bokeh')
```

```python
freq_filtered = pd.DataFrame(data = {'s1':np.random.exponential(size=1000),'s2':np.random.exponential(size=1000),})
s1='s1'
s2='s2'
```

```python
%%output backend= 'matplotlib'
p=hv.operation.histogram(hv.Points(freq_filtered, vdims=[s1,s2],kdims=[s1,s2]),dimension=s1).opts(logy=True)
hv.save(p, 'penguin_plot.png', fmt='png')
```

```python
#%%output backend= 'matplotlib'
points = hv.Points(freq_filtered, vdims = [s1,s2],kdims=[s1,s2])
xhist, yhist = (hv.operation.histogram(hv.Points(freq_filtered, vdims=[s1,s2],kdims=[s1,s2]),
                                       dimension=dim).opts() *
                hv.operation.histogram(hv.Points(freq_filtered, vdims=[s1,s2],kdims=[s1,s2]),
                                       dimension=dim).opts()
                for dim in [s1,s2])
composition = (points) << yhist.opts(width=125) << xhist.opts(height=125)
composition
```
