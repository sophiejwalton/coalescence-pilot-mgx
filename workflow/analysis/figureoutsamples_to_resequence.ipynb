```python
import pandas as pd
import numpy as np

from snp_analysis_tools_sherlock import *
from coalescence_analysis_tools import *
import iqplot
import bokeh.plotting
import bokeh.io
import holoviews as hv
from holoviews import dim, opts
import bokeh.models
from bokeh.layouts import gridplot

hv.extension('bokeh')
```

```python
fnames = glob('/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/dir_selectionv2/*')
dfs = []
for fname in fnames:
    df = pd.read_csv(fname) #.set_index(sample)
    species = fname.split('/')[-1].split('_shifts.csv')[0]
  #  df_abundance_sp = df_abundance_melted.loc[df_abundance_melted['species_id'] == int(species),:].set_index('sample')

    df['species_id'] = species
  #  df['species_abundance']= df['sample'].transform(lambda x: df_abundance_sp.loc[x, 'relative_abundance'])
   # df['Lineage'] = df_abundance_sp['Lineage'].values[0]
   # df['species'] = df_abundance_sp['species'].values[0]
    dfs.append(df)

full_df = pd.concat(dfs)

#full_df= full_df.loc[full_df['total_shift'] < .08,:]
full_df['species-mesocosm'] = full_df['species_id'] + '-' + full_df['mesocosm']
```

```python
full_df_questionable_samples = full_df.groupby(['sample','mesocosm', 'passage', 'type_meso']).median().reset_index()
full_df_questionable_samplesb = full_df_questionable_samples.loc[full_df_questionable_samples['total_shift']>=.1,:]
full_df_questionable_samplesb.sort_values(by='type_meso')
full_df_questionable_samplesb = full_df_questionable_samplesb['sample'].values
```

```python
p = iqplot.ecdf(full_df_questionable_samples , q = 'total_shift')
bokeh.io.show(p)
```
