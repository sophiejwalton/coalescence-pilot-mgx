```python
import pandas as pd
import numpy as np
import iqplot
from plot_tools import *
import itertools as it
import bokeh.plotting
import bokeh.io
import holoviews as hv
from holoviews import dim, opts
import bokeh.models
from bokeh.layouts import gridplot
from scipy.spatial.distance import jensenshannon
hv.extension('bokeh')
```

```python
df_abun = pd.read_csv('e003_round4_assembly_abundances.csv')
df_abun.loc[df_abun['relative_abundance']<1e-3,'relative_abundance']=1e-4
#df_abun_p5 = df_abun.loc[df_abun['passage'].isin([0,5])
```

```python
df_abun.loc[df_abun['exp_type']=='e003GlycerolRevival','passage']=df_abun.loc[df_abun['exp_type']=='e003GlycerolRevival','passage']+5
```

```python
sample1s = []
sample2s = []
JSDs = []
meds1=[]
meds2=[]
subs1=[]
subs2=[]
comms1=[]
comms2=[]
pass1=[]
pass2=[]
df_abun=df_abun.loc[df_abun['subject']!='AC/PP',:]
for sample1,sample2 in it.combinations(df_abun['sample'].unique(),2):
    sample1s.append(sample1)
    sample2s.append(sample2)
    s1_abuns = df_abun.loc[df_abun['sample'] == sample1,:].sort_values(by='species_id')
    
    s2_abuns  = df_abun.loc[df_abun['sample'] == sample2,:].sort_values(by='species_id')
    JSDs.append(jensenshannon(s1_abuns['relative_abundance'].values, s2_abuns['relative_abundance'].values))
```

```python
df_pairwise_JSD = pd.DataFrame(data = {'sample1': sample1s, 'sample2': sample2s, 
                                       'JSD': JSDs, })

df_pairwise_JSD['media1'] = df_pairwise_JSD['sample1'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'media'].values[0])
df_pairwise_JSD['media2'] = df_pairwise_JSD['sample2'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'media'].values[0])

df_pairwise_JSD['subject1'] = df_pairwise_JSD['sample1'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'subject'].values[0])
df_pairwise_JSD['subject2'] = df_pairwise_JSD['sample2'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'subject'].values[0])

df_pairwise_JSD['passage1'] = df_pairwise_JSD['sample1'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'passage'].values[0])
df_pairwise_JSD['passage2'] = df_pairwise_JSD['sample2'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'passage'].values[0])

df_pairwise_JSD['comm1'] = df_pairwise_JSD['sample1'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'comm'].values[0])
df_pairwise_JSD['comm2'] = df_pairwise_JSD['sample2'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'comm'].values[0])
```

```python
df_pairwise_JSD['mesocosm1'] = df_pairwise_JSD['sample1'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'mesocosm'].values[0])
df_pairwise_JSD['mesocosm2'] = df_pairwise_JSD['sample2'].transform(lambda x: df_abun.loc[df_abun['sample'] == x,'mesocosm'].values[0])
```

```python
def get_pseudo_time(x):
    if x == 1:
        return 6
    return x

df_pairwise_JSD['pseudo_time_passage1'] = df_pairwise_JSD['passage1']#.transform(get_pseudo_time)
df_pairwise_JSD['pseudo_time_passage2'] = df_pairwise_JSD['passage2']#.transform(get_pseudo_time)
```

```python
df_pairwise_JSD['same_media'] = df_pairwise_JSD['media1'] ==  df_pairwise_JSD['media2'] 
df_pairwise_JSD['same_subject'] = df_pairwise_JSD['subject1'] ==  df_pairwise_JSD['subject2'] 
df_pairwise_JSD['same_comm']=df_pairwise_JSD['comm1']==df_pairwise_JSD['comm2']
df_pairwise_JSD['same_media_same_subject'] = df_pairwise_JSD['same_media'].astype(str) + '-' + df_pairwise_JSD['same_subject'].astype(str) 
df_pairwise_JSD['passage1-passage2'] =  df_pairwise_JSD['passage1'].astype(str) + '-' + df_pairwise_JSD['passage2'].astype(str)
df_pairwise_JSD=df_pairwise_JSD.loc[(df_pairwise_JSD['subject1']!='AC/PP')*(df_pairwise_JSD['subject2']!='AC/PP'),:]
df_pairwise_JSD_p5 = df_pairwise_JSD.loc[df_pairwise_JSD['passage1-passage2'] == '5-5',:]
df_pairwise_JSD_p5['same_media_str']='Diff Media'
df_pairwise_JSD_p5.loc[df_pairwise_JSD_p5['same_media'],'same_media_str']='Same Media'

df_pairwise_JSD_p5['same_subject_str']='Diff Subject'
df_pairwise_JSD_p5.loc[df_pairwise_JSD_p5['same_subject'],'same_subject_str']='Same Subject'
df_pairwise_JSD_p5['sub-media']=df_pairwise_JSD_p5['same_subject_str'] + '-'+df_pairwise_JSD_p5['same_media_str'] 

```

```python
def get_median(array1,array2):
    return np.median(array1)-np.median(array2)

import scipy.stats 
df_pairwise_JSD_p5.columns.values
c1s = []
c2s = []
pvalues = []
stats = []
import itertools 
for pairs in itertools.combinations(df_pairwise_JSD_p5['sub-media'].unique(),2):
    print(pairs)
    c1,c2 = pairs
    x = df_pairwise_JSD_p5.loc[df_pairwise_JSD_p5['sub-media']==c1,'JSD'].values
    y = df_pairwise_JSD_p5.loc[df_pairwise_JSD_p5['sub-media']==c2,'JSD'].values
    c1s.append(c1)
    c2s.append(c2)
    
    res = scipy.stats.permutation_test((x, y), get_median, permutation_type='independent',
                       n_resamples=10000, alternative='two-sided')
    pvalues.append(res.pvalue)
    stats.append(res.statistic)
qvals = scipy.stats.false_discovery_control(pvalues)
df = pd.DataFrame({'compare1':c1s,'compare2':c2s,
                   'pval':pvalues ,'stats':stats,'qvals':qvals})
df['sig']=False
df.loc[df['qvals']<.05,'sig']='*'
df.loc[df['qvals']<.001,'sig']='**'
dfb = df.loc[(df['compare1']=='Same Subject-Same Parent Media - Same Coal Media')+(df['compare2']=='Same Subject-Same Parent Media - Same Coal Media'),
:]
dfb['qvals']=scipy.stats.false_discovery_control(dfb['pval'])
dfb.loc[dfb['qvals']<.001,'sig']='**'
df
```
