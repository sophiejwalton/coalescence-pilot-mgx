```python
def get_cmap_dict(rel_abundance, family, palette, cmap_dict = {}):
    rel_abundance_f = rel_abundance.loc[rel_abundance['family'] == family,:]
    good_sp = list(rel_abundance_f.groupby(['species']).max().sort_values(by='relative_abundance',ascending=False).index.values)
   # print(good_sp)
    if 's__' in good_sp:
        good_sp.remove('s__')
    good_sp = good_sp[:len(palette)-1]
   # print(np.sort(good_sp))
    for i, sp in enumerate(np.sort(good_sp)[::-1]):
        if sp == 's__':
            continue
        cmap_dict[f'{family}-{sp}']=palette[i]
        
    cmap_dict[f'{family}-s__'] = palette[-1]
    return cmap_dict,good_sp[:len(palette)-1]


def get_cmap_dict_linear(rel_abundance, family, primary_color, cmap_dict = {}):
    rel_abundance_f = rel_abundance.loc[rel_abundance['family'] == family,:]
    good_sp = rel_abundance_f.groupby(['species']).max().sort_values(by='relative_abundance',ascending=False)

    good_sp = list(good_sp.loc[good_sp['relative_abundance']>.01,:].index.values)
    
    if 's__' in good_sp:
        good_sp.remove('s__')
    
    print(len(good_sp))
    if len(good_sp) == 0:
        cmap_dict[f'{family}-s__'] = primary_color
        return cmap_dict,good_sp 
   # print(good_sp)
    palette = bokeh.palettes.varying_alpha_palette(primary_color, len(good_sp)+1,
                                                  100, 255)[::-1]
   # print(np.sort(good_sp))
    for i, sp in enumerate(np.sort(good_sp)):
        if sp == 's__':
            continue
        cmap_dict[f'{family}-{sp}']=palette[i+1]
        
    cmap_dict[f'{family}-s__'] = palette[0]
    return cmap_dict,good_sp


cmap_dict,good_sp= get_cmap_dict(df_abundance_melted_coal, 'f__Bacteroidaceae', bokeh.palettes.PuRd[9])


cmap_dict = {}
good_sp=[]
for family in list(family_colors.keys()):
    cmap_dict,good_sp_f = get_cmap_dict_linear(df_abundance_melted_coal, family, family_colors[family],
                             cmap_dict = cmap_dict)
    good_sp = good_sp + good_sp_f
    
good_fam = list(family_colors.keys())

```
