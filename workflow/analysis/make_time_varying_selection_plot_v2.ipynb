```python
def get_pred_sel(ts,sel_coeff, start_value,):
    pred=np.exp(ts*sel_coeff)
    pred = pred*start_value/(1 - start_value + start_value*pred)
    return pred 
    
```

```python
def get_big_df_time_varying_sel(sp,ino,t0,t1,thresh=1e-3):
    sel_stuff, df_both_meta_good = get_selection_stuff_both(sp,ino)
    sel_stuff_t0t1 = sel_stuff.loc[(sel_stuff['passage1']==t0)*(sel_stuff['passage2']==t1),:]
    sel_stuff_67 = sel_stuff.loc[(sel_stuff['passage1']==6)*(sel_stuff['passage2']==7),:]
    dfs = []
    for good_meso in sel_stuff['type_mesocosm'].unique():
        df_both_meta_good_meso= df_both_meta_good.loc[(df_both_meta_good['type_mesocosm'] == good_meso)+(df_both_meta_good['is_inoculumn']),:]
        sel_stuff_meso=sel_stuff_t0t1.loc[sel_stuff_t0t1['type_mesocosm']==good_meso,:]
        sel_stuff_meso67=sel_stuff_67.loc[sel_stuff_67['type_mesocosm']==good_meso,:]
        sel_stuff_meso=sel_stuff_meso.set_index('sample2')
                    
        df_both_meta_good_meso=df_both_meta_good_meso.set_index('sample')
        sel_stuff_meso['start_value']=df_both_meta_good_meso.loc[sel_stuff_meso.index.values,'boot_med1']
            
        sel_stuff_meso=sel_stuff_meso.reset_index()
        sel_stuff_meso=sel_stuff_meso.set_index('sample1')
        sel_stuff_meso['start_value1']=df_both_meta_good_meso.loc[sel_stuff_meso.index.values,'boot_med1']
        sel_stuff_meso=sel_stuff_meso.reset_index()
        sel_stuff_meso = sel_stuff_meso.set_index('mesocosm')
        df_both_meta_good_meso=df_both_meta_good_meso.reset_index()
        sel_stuff_meso['actual_p7']= np.nan
        #sel_stuff_meso['actual_p6']= np.nan
        if len(sel_stuff_meso)==0:
            continue 
        p7_df = df_both_meta_good_meso.loc[(df_both_meta_good_meso['passage']==7)*(df_both_meta_good_meso['type_mesocosm']==good_meso),:].set_index('mesocosm')
        p6_df = df_both_meta_good_meso.loc[(df_both_meta_good_meso['passage']==6)*(df_both_meta_good_meso['type_mesocosm']==good_meso),:].set_index('mesocosm')
        p7_mesos = np.intersect1d(sel_stuff_meso.index.values,p7_df.index.values)
        sel_stuff_meso67 = sel_stuff_meso67.set_index('mesocosm')
        p67_mesos = np.intersect1d(sel_stuff_meso.index.values,sel_stuff_meso67.index.values)
        
        #print(sel_stuff_meso.index.values, p7_df.index.values)
        sel_stuff_meso.loc[p7_mesos,'actual_p7']= p7_df.loc[p7_mesos,'boot_med1']
        sel_stuff_meso.loc[p67_mesos,'actual_p6']= p6_df.loc[p67_mesos,'boot_med1']
        sel_stuff_meso.loc[p67_mesos,'sel_med67']= sel_stuff_meso67.loc[p67_mesos,'sel_med']
        sel_stuff_meso['p7_preds'] = get_pred_sel(7-t1,sel_stuff_meso['sel_med'].values, sel_stuff_meso['start_value'].values)
        sel_stuff_meso.loc[sel_stuff_meso['actual_p6']<=thresh,'sel_med67']=sel_stuff_meso.loc[sel_stuff_meso['actual_p6']<=thresh,'sel_med']
        sel_stuff_meso.loc[sel_stuff_meso['actual_p6']>=1-thresh,'sel_med67']=sel_stuff_meso.loc[sel_stuff_meso['actual_p6']>=1-thresh,'sel_med']
        
        sel_stuff_meso_meds = sel_stuff_meso.groupby(['passage1','passage2','type_mesocosm']).median(numeric_only=True).reset_index().set_index('type_mesocosm')
        sel_stuff_meso_maxs = sel_stuff_meso.groupby(['passage1','passage2','type_mesocosm']).max(numeric_only=True).reset_index().set_index('type_mesocosm')
        sel_stuff_meso_mins = sel_stuff_meso.groupby(['passage1','passage2','type_mesocosm']).min(numeric_only=True).reset_index().set_index('type_mesocosm')

        for column in ['sel_med','actual_p7','p7_preds','sel_med67']:
            sel_stuff_meso_meds.loc[sel_stuff_meso_meds.index.values,f'max_{column}']=sel_stuff_meso_maxs.loc[sel_stuff_meso_meds.index.values,column]
            sel_stuff_meso_meds.loc[sel_stuff_meso_meds.index.values,f'min_{column}']=sel_stuff_meso_mins.loc[sel_stuff_meso_meds.index.values,column]
        dfs.append(sel_stuff_meso_meds.reset_index())
    if len(dfs)>0:
        dfs = pd.concat(dfs,axis=0)
    return dfs
             
```

```python
def make_big_plot(meds,logit_scale=False,sel_range = (-.5,.5),line=True,width=400):
    if logit_scale:
        meds = meds.copy()
        for freq_column in ['actual_p7', 'min_actual_p7',  'max_actual_p7', 'min_p7_preds','max_p7_preds','p7_preds']:
            meds[freq_column]=np.log(meds[freq_column]/(1-meds[freq_column]))
    scatter = hv.Scatter(meds, kdims =  'p7_preds', vdims=['actual_p7',hv.Dimension('sel_med',range=sel_range)]).opts(#size=4, 
                                                                                        alpha = 1.0, color = 'sel_med',#height=700,
                                                                                           width = width,line_color='black',
                                                                                                      
                                                                                      cmap = 'Magma',
                                                                                                            colorbar=True, 
                                                                                                  # logx=True, logy=True,
                                                                                                    legend_position='right',
                                                                                                        size=10,
                                                                                         )
    seg = hv.Segments(meds, ['p7_preds', 
                             'min_actual_p7', 
                             'p7_preds', 'max_actual_p7', ]).opts(color='black',alpha=.2)
    
    seg2 = hv.Segments(meds, ['min_p7_preds', 
                             'actual_p7', 
                             'max_p7_preds', 'actual_p7', ]).opts(color='black',alpha=.2)

    exps_adjust = np.array([.001,.01,.1,.5,.9,.99,.999])
    lins_adjust = np.log(exps_adjust/(1-exps_adjust))
    #exps_adjust = exps_adjust[::2]
   # lins_adjust = lins_adjust[::2]
    if logit_scale:
        scatter.opts(xticks=[(lins_adjust[i], exps_adjust[i]) for i in range(len(lins_adjust))],
                 yticks=[(lins_adjust[i], exps_adjust[i]) for i in range(len(lins_adjust))]
                )
    
    ticks = np.linspace(sel_range[0],sel_range[1],6)
    labels = [str(tick) for tick in ticks]
   # if sel_range[0]!=0:
    labels[0] = '<' + labels[0]
   # if sel_range[-1]!= 0:
    labels[-1] = '>' + labels[-1]

    #labels[-1] = '>' + labels[-1]
    print(labels)
  #  labels = ['Very Low', 'Low', 'Medium', 'High', 'Very High']
    
    # Apply colorbar customization
    scatter.opts(
        colorbar=True,
        colorbar_opts={
            'major_label_overrides': dict(zip(ticks, labels))
        }
    )

    scatter = hv.render(scatter*seg*seg2)
    if line:
        if logit_scale:
            scatter.line(np.linspace(-8,8), np.linspace(-8,8),color = 'black', )
        else:
            scatter.line(np.linspace(1e-20,1), np.linspace(1e-20,1),color = 'black', )
        
    scatter.legend.visible = True
    scatter.xaxis.axis_label = 'Predicted final frequency'
    scatter.yaxis.axis_label = 'Actual final frequency'
    #bokeh.io.show(scatter)
    return scatter,meds
```

```python
def polarize(df):
    to_polarize = df.loc[df['p7_preds']>.5,:].index.values
    for freq_column in ['actual_p7', 'min_actual_p7',  'max_actual_p7', 'min_p7_preds','max_p7_preds',]:
        df.loc[to_polarize,freq_column]= 1- df.loc[to_polarize,freq_column]
    for sel_column in ['sel_med','max_sel_med','min_sel_med']:
        df.loc[to_polarize,sel_column]= - df.loc[to_polarize,sel_column]
    df.loc[to_polarize,'p7_preds']=1-df.loc[to_polarize,'p7_preds']
    return df


def adjust(df,thresh=1e-3):
   # to_polarize = df.loc[df['pred_p7s']>.5,:].index.values
    for freq_column in ['actual_p7', 'min_actual_p7',  'max_actual_p7', 'min_p7_preds','max_p7_preds','p7_preds']:
        df.loc[df[freq_column]>1-thresh,freq_column]= 1-thresh
        df.loc[df[freq_column]<thresh,freq_column]= thresh
    return df
 
    
```
