```python
df_full = pd.concat(boths)
all_conf_ints = pd.concat([df_full[['conf_int1']],df_full[['conf_int2']].rename(columns={'conf_int2':'conf_int1'})])
all_conf_ints 
```

```python

def ecdf_transform(data):
    return 1- data.rank(method="first") / len(data)
#BEg = BE.loc[BE['relative_abundance']>1e-4,:].sort_values(by='relative_abundance',ascending=False)
all_conf_ints.loc[:, "conf int ECDF"] = all_conf_ints[
    "conf_int1"
].transform(ecdf_transform)

p1=hv.Scatter(
    data=all_conf_ints,
    kdims='conf_int1',
    vdims=[('conf int ECDF', 'ECDF'), ],
).opts(width=200,height=200, color= 'grey',xlabel='95% Conf Interval',
       #logx=True,lim=(4e-4,2e-1),
      logx=True,xlim=(1e-5,1),
       ylabel='1-P(f>x)')
p1=hv.render(p1)

```

```python
fnames = glob('/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/track_snpsv2_ALL_bootstrapv3/*/*_count_parents.csv')
counts = []
for fname in fnames:
    if 'AC' in fname:
        continue
   # df = get_both_dfs(fname)
    df1 = pd.read_csv(fname).set_index('Unnamed: 0')
    df1columns = df1.columns.values
    df1=df1.rename(columns = {df1columns[0]:'parent1',
                              df1columns[1]:'parent2',}).max()
    counts.append(df1)
    
```
