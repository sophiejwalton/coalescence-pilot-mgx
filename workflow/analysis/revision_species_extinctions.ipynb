```python
def get_comparison_on_extinctions(collision_meso,control_meso,df_abun):
    df_abun_p7 = df_abun.loc[df_abun['passage']==7,:]
    df_abun_p0 = df_abun.loc[df_abun['passage']==0,:]
  #  df_abun_p71 = df_abun_p7.loc[df_abun_p7['mesocosm']==collision_meso,:]
   # df_abun_p72 = df_abun_p7.loc[df_abun_p7['mesocosm']==control_meso,:]
    df_abun_p71 = df_abun_p7.loc[df_abun_p7['type_mesocosm']==collision_meso,:]
    df_abun_p72 = df_abun_p7.loc[df_abun_p7['type_mesocosm']==control_meso,:]
    
    
    df_abun_p01 = df_abun_p0.loc[df_abun_p0['type_mesocosm']==collision_meso,:]
    df_abun_p02 = df_abun_p0.loc[df_abun_p0['type_mesocosm']==control_meso,:]
  #  print(len(df_abun_p71),len(df_abun_p72), len(df_abun_p01), len(df_abun_p02))
   # print(control_meso)
    #df_abun_p01 = df_abun_p0.loc[df_abun_p0['sample']==df_abun_p71['inoculumn_sample'].values[0],:]
    #df_abun_p02 = df_abun_p0.loc[df_abun_p0['sample']==df_abun_p72['inoculumn_sample'].values[0],:]
    
    present_ino_both = np.intersect1d(df_abun_p01.loc[df_abun_p01['relative_abundance']>=1e-3,'species_id'].unique(),
                                      df_abun_p02.loc[df_abun_p02['relative_abundance']>=1e-3,'species_id'].unique())
    print(len(present_ino_both))
    
    df_abun_p71.loc[df_abun_p71['relative_abundance']==0,'relative_abundance']=1e-4
    df_abun_p72.loc[df_abun_p72['relative_abundance']==0,'relative_abundance']=1e-4
    
    df_abun_p71_sp_abun = df_abun_p71[['relative_abundance','species_id']].set_index('species_id')
    df_abun_p71_sp_abun = df_abun_p71_sp_abun.rename(columns = {'relative_abundance':'collision_relative_abundance'})
    
    df_abun_p72 = df_abun_p72.set_index('species_id')
    df_abun_p72 = df_abun_p72.loc[df_abun_p71_sp_abun.index.values,:]
    df_abun_p71_sp_abun['control_relative_abundance']=df_abun_p72['relative_abundance'].values
  #  df_abun_p71_sp_abun
    
    
    df_abun_p71_sp_abun = df_abun_p71_sp_abun.loc[present_ino_both,:]
    
    df_abun_p71_sp_abun['extinct'] = 'None'
    
    df_abun_p71_sp_abun.loc[(df_abun_p71_sp_abun['control_relative_abundance']>=1e-3)\
            *(df_abun_p71_sp_abun['collision_relative_abundance']<1e-3),'extinct']='collision'
    
    df_abun_p71_sp_abun.loc[(df_abun_p71_sp_abun['control_relative_abundance']<1e-3)\
            *(df_abun_p71_sp_abun['collision_relative_abundance']>=1e-3),'extinct']='control'
    
    df_abun_p71_sp_abun.loc[(df_abun_p71_sp_abun['control_relative_abundance']<1e-3)\
            *(df_abun_p71_sp_abun['collision_relative_abundance']<1e-3),'extinct']='both'
    
    df_abun_p71_sp_abun['counts']=1
    

    return df_abun_p71_sp_abun.groupby(['extinct']).sum(numeric_only=True).reset_index()
```

```python
collisions_list = []
controls_list = []
boths_list = []
collision_mesos = []
control_mesos = []
type_mesos = ['AA-AE-mGAM-mGAM','AA-AF-mGAM-mGAM',  'AE-AF-mGAM-mGAM',
              'AA-AE-mBHI-mBHI','AA-AF-mBHI-mBHI', 'AE-AF-mBHI-mBHI',
             'AA-AA-mGAM-mGAM','AF-AF-mGAM-mGAM','AE-AE-mGAM-mGAM',
             'AA-AA-mBHI-mBHI','AF-AF-mBHI-mBHI','AE-AE-mBHI-mBHI']
#df_abun_good = df_abun_p7.loc[df_abun_p7['type_mesocosm'].isin(type_mesos),:]
df_abun_good = df_abun_p7_med.loc[df_abun_p7_med['type_mesocosm'].isin(type_mesos),:]
good_mesos = ['AA-AE-mGAM-mGAM','AA-AF-mGAM-mGAM',  'AE-AF-mGAM-mGAM',
              'AA-AE-mBHI-mBHI','AA-AF-mBHI-mBHI', 'AE-AF-mBHI-mBHI',
            ]
#for collision_meso in df_abun_good['mesocosm'].unique(): 
for collision_meso in good_mesos:
  #  _,sub1,sub2,med1,med2 = collision_meso.split('-')
    sub1,sub2,med1,med2 = collision_meso.split('-')
    control_type_mesos = [f'{sub1}-{sub1}-{med1}-{med2}',f'{sub2}-{sub2}-{med1}-{med2}']
    for control_meso in control_type_mesos:
        if control_meso == 'AF-AF-mBHI-mBHI': # this one looks weird
            continue 
        info = get_comparison_on_extinctions(collision_meso,control_meso,df_abun_med).reset_index()
     #   print(info)
        collisions = 0
        if len(info.loc[info['extinct']=='collision','counts'])>0:
            collisions = info.loc[info['extinct']=='collision','counts'].values[0]
        
        controls = 0
        if len(info.loc[info['extinct']=='control','counts'])>0:
            controls = info.loc[info['extinct']=='control','counts'].values[0]
        both = 0
        if len(info.loc[info['extinct']=='both','counts'])>0:
            both = info.loc[info['extinct']=='both','counts'].values[0]
        collision_mesos.append(collision_meso)
        control_mesos.append(control_meso)
        controls_list.append(controls)
        collisions_list.append(collisions)
        boths_list.append(both)
```

```python
df = pd.DataFrame(data = {'collision_meso':collision_mesos,
                          'control_meso':control_mesos,
                          'collision_only_extinctions':collisions_list,
                          'control_only_extinctions':controls_list,
                          'both_extinctions':boths_list,})
df['fract_collision_ext_unique']=df['collision_only_extinctions']/(df['collision_only_extinctions']+df['both_extinctions'])

df['fract_control_ext_unique']=df['control_only_extinctions']/(df['control_only_extinctions']+df['both_extinctions'])

df['fract_shared_collision']=df['both_extinctions']/(df['collision_only_extinctions']+df['both_extinctions'])

df['fract_shared_control']=df['both_extinctions']/(df['control_only_extinctions']+df['both_extinctions'])

df['media']=df['control_meso'].transform(lambda x: x.split('-')[-1])
```

```python
import numpy as np
def statistic(x, y, axis):
    return np.mean(x, axis=axis) - np.mean(y, axis=axis)
```

```python
from scipy.stats import permutation_test
# because our statistic is vectorized, we pass `vectorized=True`
# `n_resamples=np.inf` indicates that an exact test is to be performed
x = df['unique_collision'].values
y = df['unique_control'].values
res = permutation_test((x, y), statistic, vectorized=True,
                       n_resamples=100000, alternative='greater')
print(res.statistic)
print(res.pvalue)
```
