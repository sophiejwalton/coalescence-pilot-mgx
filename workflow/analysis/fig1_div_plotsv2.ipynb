```python
def permutation_test(zeros,ones,twos,threes,bsreps = 1000):
    total_comps = (zeros+ones+twos+threes)*3
    comps = zeros+ones+twos+threes
    total_oligo= int(ones+twos*2+threes*3)
    ones_bs=np.zeros(bsreps)
    twos_bs = np.zeros(bsreps)
    threes_bs=np.zeros(bsreps)
    for i in range(bsreps):
        rep = np.zeros(total_comps)
        random_inds = np.random.choice(np.arange(total_comps),size=total_oligo)
        rep[random_inds]=1
        rep = np.reshape(rep,(comps,3))
        rep=rep.sum(axis=1)
        ones_bs[i]=np.sum(rep==1)
        twos_bs[i]=np.sum(rep==2)
        threes_bs[i]=np.sum(rep==3)
    return ones_bs,twos_bs,threes_bs
        
def permutation_test_twos(zeros,ones,twos,bsreps = 1000):
    total_comps = (zeros+ones+twos)*2
    comps = zeros+ones+twos
    total_oligo= int(ones+twos*2)
    ones_bs=np.zeros(bsreps)
    twos_bs = np.zeros(bsreps)
    threes_bs=np.zeros(bsreps)
    for i in range(bsreps):
        rep = np.zeros(total_comps)
        random_inds = np.random.choice(np.arange(total_comps),size=total_oligo)
        rep[random_inds]=1
        rep = np.reshape(rep,(comps,2))
        rep=rep.sum(axis=1)
        ones_bs[i]=np.sum(rep==1)
        twos_bs[i]=np.sum(rep==2)
    return ones_bs,twos_bs


        
def permutation_test_twos_threes(zeros_two,ones_two,twos_two,
                                 zeros_three, ones_three, twos_three,threes_three,
                                 bsreps = 1000):
    total_comps2 = (zeros_two+ones_two+twos_two)*2
    total_comps3 = (zeros_three+ones_three+twos_three+threes_three)*3 
 #   print(total_comps3,zeros_three,ones_three,twos_three, threes_three)
    
    comps2 = zeros_two+ones_two+twos_two
    comps3=zeros_three+ones_three+twos_three + threes_three
   
    total_oligo= int(ones_two+twos_two*2) + int(ones_three+twos_three*2+threes_three*3)
    ones_two_bs=np.zeros(bsreps)
    twos_two_bs = np.zeros(bsreps)
    ones_three_bs=np.zeros(bsreps)
    twos_three_bs=np.zeros(bsreps)
    threes_three_bs=np.zeros(bsreps)
    for i in range(bsreps):
        rep = np.zeros(total_comps2+total_comps3)
        random_inds = np.random.choice(np.arange(total_comps2+total_comps3),size=total_oligo)
        rep[random_inds]=1
        rep2 = rep[:total_comps2]
      #  print(total_comps2,total_comps3)
        rep2 = np.reshape(rep2,(comps2,2))
        rep3 = rep[total_comps2:]
        
        rep3 = np.reshape(rep3,(comps3,3))
        
        rep2=rep2.sum(axis=1)
        rep3 = rep3.sum(axis=1)
        ones_two_bs[i]=np.sum(rep2==1)
        twos_two_bs[i]=np.sum(rep2==2)

        ones_three_bs[i]=np.sum(rep3==1)
        twos_three_bs[i]=np.sum(rep3==2)
        threes_three_bs[i]=np.sum(rep3==3)
    return twos_two_bs+threes_three_bs+twos_three_bs
            
            
```
