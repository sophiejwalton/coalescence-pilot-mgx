```python
def make_div_plot(df,color='black',alpha=1.,p=None,log=True):
    df = df.sort_values(by='passage')
    if p is None:
        p = bokeh.plotting.figure(width = 500, height = 250,  y_axis_type="log")
        p.quad(top=[1], bottom=[1e-6], left=[6],
       right=[15], color=bokeh.palettes.Light[3][0],alpha=.2)
    p.line(df['passage'].values,
                   df['div'].values,color=color,alpha=alpha)
    p.scatter(df['passage'].values,
                   df['div'].values,color = color,size = 5,
                      alpha=alpha)
    p.yaxis.axis_label = f'SNP Diversity'
    p.xaxis.axis_label = 'Passage'
   # p2.y_range = bokeh.models.Range1d(-.005,1.005)
   # p2.x_range = bokeh.models.Range1d(-.5,7.5)
    p.xgrid.grid_line_color = None
    p.ygrid.grid_line_color = None
    p.y_range = bokeh.models.Range1d(1e-5,1)
    p.x_range = bokeh.models.Range1d(-.5,14.5)
    p.title.text_font_style='italic'
    p.yaxis.axis_label_text_font_style = 'normal'
    p.xaxis.axis_label_text_font_style = 'normal'
    label_font_size = '22px'
    tick_font_size = '15px'
    p.xaxis.axis_label_text_font_size=label_font_size
    p.xaxis.major_label_text_font_size=tick_font_size
    p.yaxis.axis_label_text_font_size=label_font_size
    p.yaxis.major_label_text_font_size=tick_font_size
    p.xaxis.axis_label = 'Timepoint'
    p.xaxis.minor_tick_line_color= None
  #  p.yaxis.minor_tick_line_color= None
    p.title.text_font_style='italic'
    p.yaxis.axis_label='% Polymorphic SNPs'
    p.title.text_font_size=label_font_size
    p.ray(x=-2,y=1e-3, color='grey',alpha=.2,width=5)
    p.y_range = bokeh.models.Range1d(5e-6,5e-2)
    
    return p
```
