```python
def get_pred_sel(ts,sel_coeff, start_value,):
    pred=np.exp(ts*sel_coeff)
    pred = pred*start_value/(1 - start_value + start_value*pred)
    return pred 

    
```

```python
xs = df['het_pred'].values
ys = df['het_act'].values

def get_phetstar(xs,ys,hetstar):
    ys1 = ys[xs>=hetstar]
    xs1 = xs[xs>=hetstar]
    good1 = np.sum(xs1<ys1)
    p1 = scipy.stats.binom.sf(good1-1, len(xs1),1/2)
    
    ys2 = ys[xs<hetstar]
    xs2 = xs[xs<hetstar]
    good2 = np.sum(xs2<ys2)
    p2 = scipy.stats.binom.sf(good2-1, len(xs2),1/2)

    return np.min([p1,p2])

pstar = get_phetstar(xs,ys,0)
hets = np.linspace(0,.25,10)

def get_phat(xs,ys, hets = np.linspace(0,.25,10)):
    ps = []
    
    for het in hets:
        ps.append(get_phetstar(xs,ys,het))
    return np.min(ps), hets[np.argmin(ps)], ps

def get_phat(xs,ys, hets = np.linspace(0,.25,10)):
    ps = []
    
    for het in hets:
        ps.append(get_phetstar(xs,ys,het))
    return np.min(ps), hets[np.argmin(ps)], ps
    
phat,hetstar, ps = get_phat(xs,ys, hets = hets)
phat

def get_bs_reps(xs, hets,bsreps = 10000):
    rep_ps = np.zeros(bsreps)
    rep_het = np.zeros(bsreps)
    for rep in range(bsreps):
        ys_rand = np.random.binomial(1, 1/2,size = len(xs))
        ys = xs + -.01*(1- ys_rand) + .01*(ys_rand)
        pbs,hbs,_ = get_phat(xs,ys, hets = hets)
        rep_ps[rep]= pbs
        rep_het[rep]=hbs
    return rep_ps, rep_het
        
xs = df['het_pred'].values
ys = df['het_act'].values

hets = np.linspace(.0,.25,200)
rep_ps, rep_het = get_bs_reps(xs, hets,bsreps = 10000)

        
phat,hetstar, ps = get_phat(xs,ys, hets = hets)
```

```python
def get_bs_reps_for_plot(xs, hets,bsreps = 1000):
    lower_conf = np.zeros(len(hets))
    upper_conf = np.zeros(len(hets))
    
    for i,het in enumerate(hets):
        ps_reps = np.zeros(bsreps)
        for rep in range(bsreps):
                ys_rand = np.random.binomial(1, 1/2,size = len(xs))
                ys = xs + -.01*(1- ys_rand) + .01*(ys_rand)
                pbs = get_phetstar(xs,ys,het)
                ps_reps[rep]= pbs
        lower_conf[i]= np.percentile(ps_reps,2.5)
        upper_conf[i] = np.percentile(ps_reps,97.5)
    return lower_conf, upper_conf
hets = np.linspace(.0,.25,200)
lower_conf, upper_conf = get_bs_reps_for_plot(xs, hets,bsreps = 1000)
```
