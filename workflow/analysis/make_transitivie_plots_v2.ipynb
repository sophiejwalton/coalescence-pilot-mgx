```python
def transitive_relation(x,c):
    return x+c
    
def get_estimate_c(full_dfp7_goodgr_small, passages_fit = [0,1]):
    xs=[]
    ys=[]
    xs_min = []
    xs_max = []
    passages=[]
    xsfit = []
    ysfit = []
    for passage in [0,1,2,3,4,5,6,7]:
        passage_df = full_dfp7_goodgr_small.loc[full_dfp7_goodgr_small['passage']==passage,:]
        if len(passage_df['parent_subjects'].unique())<3:
            continue
        x1 = passage_df.loc[passage_df['parent_subjects']=='AA-AE','boot_med'].values[0]
        x2 =passage_df.loc[passage_df['parent_subjects']=='AE-AF','boot_med'].values[0]
        xs.append(x1+x2)
        ys.append(passage_df.loc[passage_df['parent_subjects']=='AA-AF','boot_med'].values[0])
        passages.append(passage)
        if passage in passages_fit:
            
            xsfit.append(x1+x2)
            ysfit.append(passage_df.loc[passage_df['parent_subjects']=='AA-AF','boot_med'].values[0])
    if len(ys)==0:
        return np.nan,np.nan,np.nan,np.nan
    if len(ysfit)==0:
        return np.nan,np.nan,np.nan,np.nan
    c,_ = curve_fit(transitive_relation,xsfit,ysfit)
  #  print(curve_fit(transitive_relation,xs,ys))
   # print(c)
    return c[0],xs, ys,passages


def get_estimate_c_errors(full_dfp7, passages_fit = [0,1]):
    all_dfs = []
    for i in range(100):
        x1r = np.random.choice(full_dfp7.loc[full_dfp7['parent_subjects']=='AA-AE','replicate'].unique(), len(full_dfp7.loc[full_dfp7['parent_subjects']=='AA-AE','replicate'].unique()))
      #  print(x1r)
        x1 = full_dfp7.loc[(full_dfp7['parent_subjects']=='AA-AE')*((full_dfp7['replicate'].isin(x1r)) + (full_dfp7['passage']==0)),:]
       # print(full_dfp7.loc[full_dfp7['replicate'].isin(list(x1r)),'passage'])
        x2r=np.random.choice(full_dfp7.loc[full_dfp7['parent_subjects']=='AE-AF','replicate'].unique(), len(full_dfp7.loc[full_dfp7['parent_subjects']=='AE-AF','replicate'].unique()))
        x2 = full_dfp7.loc[(full_dfp7['parent_subjects']=='AE-AF')*((full_dfp7['replicate'].isin(x2r))+ (full_dfp7['passage']==0)),:]
        yr=np.random.choice(full_dfp7.loc[full_dfp7['parent_subjects']=='AA-AF','replicate'].unique(), len(full_dfp7.loc[full_dfp7['parent_subjects']=='AA-AF','replicate'].unique()))
        y = full_dfp7.loc[(full_dfp7['parent_subjects']=='AA-AF')*((full_dfp7['replicate'].isin(yr))+(full_dfp7['passage']==0)),:]
        df = pd.concat([x1,x2,y])
       # print(x1,x2,y)
        c,xs,ys,passages = get_estimate_c(df.groupby(['parent_subjects','passage']).median(numeric_only=True).reset_index(),passages_fit=passages_fit)
        df2 = pd.DataFrame(data={'x':xs,'y':ys,'passages':passages})
        df2['c']=c
        df2['ys_est']=df2['c']+df2['x']
        all_dfs.append(df2)
    return pd.concat(all_dfs)
        
```

```python

def adjust_freq(df_both,shift=True, thresh=1e-3,):
    e003_metadata = pd.read_csv('e003_coalescence_metadata_round4_good.csv').set_index('sample')
    df_both = df_both.loc[np.intersect1d(df_both.index.values,e003_metadata.index.values),:]
    df_both=df_both.loc[df_both['shifts_med']<.1,:]

    df_both_meta_good = pd.concat([df_both,e003_metadata.loc[np.intersect1d(df_both.index.values,
                                                                                   e003_metadata.index.values),:]],
                                         axis=1).reset_index()

    #df_both_meta_good=df_both_meta.loc[df_both_meta['total_shift']<.1,:]

    for col in ['boot_med','boot_low','boot_high']:
        df_both_meta_good.loc[df_both_meta_good[col]<thresh,col]=thresh
        df_both_meta_good.loc[df_both_meta_good[col]>1-thresh,col]=1-thresh
    df_both_meta_good_logit = df_both_meta_good.copy()
    df_both_meta_good_logit[['boot_med','boot_low','boot_high']] = np.log(df_both_meta_good_logit[['boot_med','boot_low','boot_high',]]/(1-df_both_meta_good_logit[['boot_med','boot_low','boot_high',
   ]]) )

    return df_both_meta_good_logit
```
