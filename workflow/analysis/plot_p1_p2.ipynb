```python

def boot_gstats(gstats, bootstraps = 10000):
    meds = np.zeros(bootstraps)
    for b in range(bootstraps):
       # print(b)
        new_gstats= np.random.choice(gstats, size = len(gstats))
        meds[b] = np.mean(new_gstats)
    return meds
```

```python
def get_competitions(sel_period):
    all_dfs=[]
    species_list = glob('/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/track_snpsv2_sel_bootstrapv3/*/')
    species_list =[sp.split('/')[-2] for sp in species_list]
    for sp in species_list:
        for ino in ['AA-AE-mBHI','AA-AF-mBHI','AE-AF-mBHI',
                    'AA-AE-mGAM','AA-AF-mGAM','AE-AF-mGAM',]:
         #   try:
            if os.path.exists(f'/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/track_snpsv2_sel_bootstrapv3/{sp}/{ino}_parentboth_info.csv'):
              #  print('wpp')
                if not os.path.exists(f'/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/track_snpsv2_ALL_bootstrapv3_both/{sp}/{ino}_parentboth_info.csv'):
                    continue
                df,_=get_selection_stuff_both(sp,ino)
                df_freq = pd.read_csv(f'/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/track_snpsv2_ALL_bootstrapv3_both/{sp}/{ino}_parentboth_info.csv').set_index('sample')
                e003_metadata = pd.read_csv('e003_coalescence_metadata_round4_good.csv').set_index('sample')
               
                #df_both = get_both_dfs(fname1)
                df_freq = df_freq.loc[np.intersect1d(df_freq.index.values,e003_metadata.index.values),:]
                df_freq=df_freq.loc[df_freq['shifts_med']<.1,:]
       
                df_freq = pd.concat([df_freq,e003_metadata.loc[np.intersect1d(df_freq.index.values,
                                                                                   e003_metadata.index.values),:]], axis=1).reset_index()
               
                # df_freq = df_freq.loc['
                df = df.loc[(df['passage1']==sel_period[0])*(df['passage2']==sel_period[1]),:]
               # print(df.columns.values)
                df['frequencyp1']=np.nan
                df['frequencyp0']=np.nan
                
                for meso in df['mesocosm'].unique():
                    df_freq_meso = df_freq.loc[(df_freq['mesocosm'] == meso)*(df_freq['passage'] == sel_period[0]),:]
                    df.loc[df['mesocosm'] == meso,'frequencyp1']= df_freq_meso['actual_med'].values
                    ino = df_freq_meso['inoculumn_sample'].values[0]
                  #  print(ino)
                   # print( df_freq.loc[df_freq['sample']==ino,'actual_med'])
                    if len(df_freq.loc[df_freq['sample']==ino,'actual_med'].values)>0:
                        df.loc[df['mesocosm'] == meso,'frequencyp0'] = df_freq.loc[df_freq['sample']==ino,'actual_med'].values[0]

                df['species']=sp
                df['inoculumn']=ino
             #   print(df)
           # except:
            #    continue
                df['species_id']=sp
                df['species-ino']= sp+'-' + ino
            
                all_dfs.append(df)
                
                
    full_dfp7=pd.concat(all_dfs)
    full_dfp7 = full_dfp7.loc[(full_dfp7['passage1']==sel_period[0])*(full_dfp7['passage2']==sel_period[1]),:]
    
    full_dfp7 = full_dfp7.loc[~full_dfp7['parent_subjects'].isin(['AA-AC/PP','AC/PP-AE','AC/PP-AF']),:]
    full_dfp7['p7_s']=full_dfp7['sel_med1']
        
    full_dfp7['type_mesocosm'] = full_dfp7['parent_subjects'] + '-' + full_dfp7['parent_media'] \
        + '-' + full_dfp7['media']
    #full_dfp7.loc[full_dfp7['p7_s'].isna(),'p7_s'] = full_dfp7.loc[full_dfp7['p7_s'].isna(),'p5_s']
    full_dfp7 = full_dfp7.loc[~full_dfp7['p7_s'].isna(),:]
    full_dfp7['parent1'] = full_dfp7['parent_subjects'].transform(lambda x: x.split('-')[0])
    full_dfp7['winner']='parent1'
    full_dfp7['parent2'] = full_dfp7['parent_subjects'].transform(lambda x: x.split('-')[1])
    full_dfp7['loser']='parent2'
    
    full_dfp7.loc[full_dfp7['p7_s']>0,'winner'] = full_dfp7.loc[full_dfp7['p7_s']>0,'parent1'] 
    full_dfp7['winner_numeric'] = (full_dfp7['p7_s']>0).astype(int)
    full_dfp7.loc[full_dfp7['p7_s']<0,'winner'] = full_dfp7.loc[full_dfp7['p7_s']<0,'parent2'] 
    full_dfp7.loc[full_dfp7['p7_s']>0,'loser'] = full_dfp7.loc[full_dfp7['p7_s']>0,'parent2'] 
    full_dfp7.loc[full_dfp7['p7_s']<0,'loser'] = full_dfp7.loc[full_dfp7['p7_s']<0,'parent1'] 
    full_dfp7['counts']=1
    full_dfp7_gr = full_dfp7.groupby(['type_mesocosm','species_id','parent_subjects','media','inoculumn',]).median(numeric_only=True).reset_index()
    full_dfp7_gr['winner_numeric'] = full_dfp7_gr['p7_s']>0.
    full_dfp7_gr['counts']=1
    full_dfp7_gr['het'] = full_dfp7_gr['winner_numeric']*(full_dfp7_gr['winner_numeric']-full_dfp7_gr['counts'])
    full_dfp7_gr['sel_period']=str(sel_period[0])+'-'+str(sel_period[1])
    #full_dfp7_gr_Good = full_dfp7_gr.loc[full_dfp7_gr['het']==0,:]

    return full_dfp7_gr, full_dfp7
```
