```python
import pandas as pd
import numpy as np

from snp_analysis_tools_sherlock import *
from coalescence_analysis_tools import *
import iqplot
import bokeh.plotting
import bokeh.io
import holoviews as hv
from holoviews import dim, opts
import bokeh.models
from bokeh.layouts import gridplot

hv.extension('bokeh')
```

```python
fname = '~/git/coalescence-pilot-mgx/midas2_output/mergev2/species/species_relative_abundance.tsv'
df_abundance = pd.read_csv(fname, delimiter = '\t')
df_abundance
fname = '~/git/coalescence-pilot-mgx/workflow/out/midas2_output/old_species/species/metadata.tsv'
df_metadata= pd.read_csv(fname, delimiter = '\t')
df_metadata

def transform_df(df_abundance, ):
    df_abundance['Lineage'] = df_abundance['species_id'].transform(lambda x: df_metadata.loc[df_metadata['species_id'] == x,'Lineage'].values[0])
    df_abundance['species'] = df_abundance['Lineage'].transform(lambda x: x.split(';')[-1])
    df_abundance['genus'] = df_abundance['Lineage'].transform(lambda x: x.split(';')[-2])
    df_abundance['family'] = df_abundance['Lineage'].transform(lambda x: x.split(';')[-3])
    df_abundance['phyla'] = df_abundance['Lineage'].transform(lambda x: x.split(';')[1])
    return df_abundance

df_abundance_melted = pd.melt(df_abundance, 
                                  var_name = 'sample', 
                                  id_vars='species_id',
                                  value_name = 'relative_abundance')


df_abundance_melted = transform_df(df_abundance_melted)
df_metadata = transform_df(df_metadata)
df_metadata['species_id'] = df_metadata['species_id'].astype(str)
#df_metadata = df_metadata.set_index('species_id')
```

```python
fnames = glob('/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/dir_selectionavg/*')
dfs = []
for fname in fnames:
    df = pd.read_csv(fname) #.set_index(sample)
    #print(fname,df['strain_freq'].min())
    species = fname.split('/')[-1].split('_shifts.csv')[0]
    depths_df = f'/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/track_snpsavg/{species}/{species}_median_depths.csv'
    depths_df = pd.read_csv(depths_df).rename(columns = {'Unnamed: 0': 'sample', '0': 'depth'}).set_index('sample')
#    print(depths_df)
    df_abundance_sp = df_abundance_melted.loc[df_abundance_melted['species_id'] == int(species),:].set_index('sample')

    df['species_id'] = species
  #  df['species_abundance']= df['sample'].transform(lambda x: df_abundance_sp.loc[x, 'relative_abundance'])
    df['sample_depth'] = df['sample'].transform(lambda x: depths_df.loc[x,'depth'])
    df['Lineage'] = df_abundance_sp['Lineage'].values[0]
    df['species'] = df_abundance_sp['species'].values[0]
    dfs.append(df)

full_df = pd.concat(dfs)
print(full_df.loc[full_df['strain_freq']<1.,'strain_freq'].max())
#full_df['minor_strain_freq'] = full_df['strain_freq']
full_df = full_df.loc[full_df['sample_depth']>=5,:]
ep=1e-3
#full_df = full_df.loc[full_df['strain_depth']>1/ep,:]
full_df.loc[full_df['strain_freq']<=ep,'strain_freq'] = ep
full_df.loc[full_df['strain_freq']>=1-ep,'strain_freq'] = 1-ep
full_df.loc[full_df['opp_strain_freq']<=ep,'opp_strain_freq'] = ep 
full_df.loc[full_df['opp_strain_freq']>=1-ep,'opp_strain_freq'] = 1-ep


#full_df.loc[full_df['minor_strain_freq']>.5,'minor_strain_freq'] = 1. - full_df.loc[full_df['minor_strain_freq']>.5,'minor_strain_freq'] 
#full_df= full_df.loc[full_df['total_shift'] < .1,:]
full_df['species-mesocosm'] = full_df['species_id'] + '-' + full_df['mesocosm']
full_df['species_id'] = full_df['species_id'].astype(str)

full_df = transform_df(full_df)
```

```python
full_df['log_total_shift'] = np.log10(full_df['total_shift']+1e-3)
full_df = full_df.loc[~full_df['parent_subjects'].isin(['AA-AC/PP','AC/PP-AE','AC/PP-AF']),:]

frequencies, edges = np.histogram(full_df.loc[~full_df['total_shift'].isna(),'total_shift'], 30)
#print('Values: %s, Edges: %s' % (frequencies.shape[0], edges.shape[0]))
p = hv.render(hv.Histogram((edges, frequencies)).opts(width=500, height=200, xlabel='|1-(Strain 1 Freq + Strain 2 Freq)|', 
                                        ylabel='Counts',fill_color= bokeh.palettes.Bright[6][1]
                                       ))
p.ray(x=.1,y=0,angle=np.pi/2,width=2)
bokeh.io.show(p)
```
