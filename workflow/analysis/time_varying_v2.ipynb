```python
import pandas as pd
import numpy as np

from snp_analysis_tools_sherlock import *
from coalescence_analysis_tools import *
import iqplot
import bokeh.plotting
import bokeh.io
import holoviews as hv
from holoviews import dim, opts
import bokeh.models
from bokeh.layouts import gridplot

hv.extension('bokeh')
```

```python
fnames = glob('/Users/sophiewalton/git/coalescence-pilot-mgx/workflow/report/track_snpsv2_shift_self_test_13/*/*parent1_info.csv')
fnames
full_df = []
for fname in fnames:
    df = pd.read_csv(fname)
    df['species_id'] = fname.split('/')[-2]
    full_df.append(df)
full_df = pd.concat(full_df).drop(columns = 'Unnamed: 0')
#full_df_good = full_df.loc[full_df['as_extreme7']<10000,:]
#full_df_good['as_extreme'].max()
full_df.columns
```

```python
full_df['is_sig'] = full_df['diff7_med'].abs()>.1
p=hv.Scatter(full_df, kdims = ['diff7_med'],vdims = ['diff5_med', 'is_sig']).opts(color='is_sig',legend_position='bottom_right')
p
```

```python
full_dfp7 = pd.read_csv('selection_coefficients_fitting2_v2.csv')
full_dfp7  = full_dfp7.loc[full_dfp7['total_shift']<.05,:]


full_df['species-mesocosm'] = full_df['species_id'].astype(str) + '-' + full_df['mesocosms']
full_df = full_df.loc[full_df['species-mesocosm'].isin(full_dfp7['species-mesocosm'].unique()),:]
full_df['p3_s'] = full_df['species-mesocosm'].transform(lambda x: full_dfp7.loc[full_dfp7['species-mesocosm']==x,'p3_s'].values[0])
full_dfp7['p3_s_pol'] = full_dfp7['p3_s']
full_dfp7.loc[full_dfp7['p0_freq']>.5,'p3_s_pol']= -full_dfp7.loc[full_dfp7['p0_freq']>.5,'p3_s']
full_dfp7['to_repol'] = full_dfp7['p0_freq']>.5


full_df['large'] = full_df['diff_med'].abs()>.1
full_df['type_mesocosm'] = full_df['mesocosms'].transform(lambda x: '-'.join(x.split('-')[1:]))

full_df['p3_s_pol'] = full_df['species-mesocosm'].transform(lambda x: full_dfp7.loc[full_dfp7['species-mesocosm']==x,'p3_s_pol'].values[0])
full_df['p3_freq'] = full_df['species-mesocosm'].transform(lambda x: full_dfp7.loc[full_dfp7['species-mesocosm']==x,'p3_freq'].values[0])
full_df['total_shift'] = full_df['species-mesocosm'].transform(lambda x: full_dfp7.loc[full_dfp7['species-mesocosm']==x,'total_shift'].values[0])
full_df['act7_pol'] =  full_df['act_7'].copy()
full_df['to_repol']=full_df['species-mesocosm'].transform(lambda x: full_dfp7.loc[full_dfp7['species-mesocosm']==x,'to_repol'].values[0])
full_df.loc[full_df['to_repol'],'act7_pol'] = 1- full_df.loc[full_df['to_repol'],'act_7']
full_df['AC'] = full_df['mesocosms'].transform(lambda x: 'AC' in x)
full_df= full_df.loc[~full_df['AC'],:]

full_dfp7no3 = full_dfp7.loc[full_dfp7['p3_freq']>1e-2,:]
full_dfp7no3 = full_dfp7no3.loc[full_dfp7no3['p3_freq']<1-1e-2,:]

full_df_no3 = full_df.loc[full_df['species-mesocosm'].isin(full_dfp7no3['species-mesocosm']),:]
```
