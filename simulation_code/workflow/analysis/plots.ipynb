```python
def export_plot_pdf(p, fname_prefix):
    p.output_backend = "svg"
    plot_fname_svg=f'{fname_prefix}.svg'
    plot_fname_pdf=f'{fname_prefix}.pdf'
    bokeh.io.export_svg(p,filename=plot_fname_svg)
    convert_scripts = f'rsvg-convert -f pdf -o {plot_fname_pdf} {plot_fname_svg}' 
    os.system(convert_scripts)
```

```python
def make_nice_plot(df, marker):
    p = hv.Points(df,vdims=['logit_input',hv.Dimension('depth_infer_full',range = (0,30))],kdims=['logit_input','logit_est'],
                          ).opts(color='depth_infer_full',colorbar=True, #cmap=bokeh.palettes.Light[8],
                                                                                       size=5,alpha=.5,xlabel='Input Abundance',
                                            marker = marker,
                                           ylabel='Estimated Abundance', show_legend=False)
    exps_adjust = np.array([1e-5, .0001, .001,.01,.1,.5,.9,.99,.999,.9999, 1-1e-5])
        
    lins_adjust = np.log(exps_adjust/(1-exps_adjust))

    p.opts(xticks=[(lins_adjust[i], exps_adjust[i]) for i in range(len(lins_adjust))],
                         yticks=[(lins_adjust[i], exps_adjust[i]) for i in range(len(lins_adjust))],
               width=580,height=350, xlim=(-12,12),ylim=(-12,12),#ylabel='Est',xlabel='Input',
               legend_position='bottom_right',size=8,)
    return p 
    
```

```python
from scipy.stats import permutation_test

def statistic_mean(x, y):
    return np.median(x) - np.median(y)
x=df_full_nonzero.loc[df_full_nonzero['genome-sample']=='genome1-sample1','bias'].values
y=df_full_nonzero.loc[df_full_nonzero['genome-sample']=='genome1-sample2','bias'].values
res = permutation_test((x, y), statistic_mean,
                       n_resamples=100000, alternative='two-sided')

print(res.pvalue)

x=df_full_nonzero.loc[df_full_nonzero['genome-sample']=='genome2-sample1','bias'].values
y=df_full_nonzero.loc[df_full_nonzero['genome-sample']=='genome2-sample2','bias'].values
res = permutation_test((x, y), statistic_mean,
                       n_resamples=100000, alternative='two-sided')

print(res.pvalue)
```
